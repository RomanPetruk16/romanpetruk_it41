```python
rows = np.array([
    temps,
    days_in_month
])

print(rows)
print("shape:", rows.shape)
```

```python
correction = np.array([1, 0])

print(correction)
print("shape:", correction.shape)
```

```python
rows + correction
```

```python
correction_reshaped = correction.reshape(2, 1)

print(correction_reshaped)
print("shape:", correction_reshaped.shape)
```

```python
result = rows + correction_reshaped

print("Результат:")
print(result)
```

1. Різниця між shape == (12,) і shape == (1, 12)

(12,) — це одновимірний масив із 12 елементів. (1, 12) — це двовимірний масив, який має 1 рядок і 12 стовпців. Вони мають однакову кількість елементів, але різну кількість вимірів.

2. Чому temps[temps > 15] повертає масив іншої довжини?

temps > 15 створює булеву маску. При булевій індексації залишаються тільки ті елементи, для яких умова має значення True. Тому результат містить тільки температури вище 15°C і може бути коротшим за початковий масив із 12 елементів.

3. Що станеться при додаванні (3,) до (3, 12)?

Виникне помилка broadcasting, тому що NumPy порівнює розміри справа наліво. Розміри 12 і 3 несумісні. Щоб виконати додавання одного значення до кожного рядка, масив (3,) потрібно перетворити на (3, 1).

4. Чому rows + correction дає помилку, а rows + correction.reshape(2, 1) працює?

rows має форму (2, 12), а correction має форму (2,). Розміри 12 і 2 несумісні, тому виникає помилка. Після reshape(2, 1) форми стають (2, 12) і (2, 1). Розмір 1 може бути розширений до 12, тому broadcasting дозволяє виконати операцію.

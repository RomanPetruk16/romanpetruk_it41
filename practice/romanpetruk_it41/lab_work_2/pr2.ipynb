```python
def get_season(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    elif month_number in [3, 4, 5]:
        return "весна"
    elif month_number in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"
```

```python
df["season"] = df.index.to_series().apply(get_season)
```

## Письмові відповіді

### 1. Чому `.loc[1:6]` і `.iloc[0:6]` можуть повернути різну кількість рядків?

`.loc` виконує відбір за підписами індексу, а `.iloc` — за позиціями рядків. 
Тому результат залежить від того, які саме значення має індекс DataFrame. 
Крім того, `.loc[1:6]` включає кінцевий підпис 6, а `.iloc[0:6]` не включає позицію 6.

### 2. Чому `and` викликає помилку, а `&` працює?

У Pandas вирази `df["avg_temp"] > 10` та `df["precip"] > 50` повертають цілі серії значень `True` і `False`. 
Оператор `and` призначений для роботи з окремими булевими значеннями, тому для Series він не підходить. 
Оператор `&` виконує поелементне логічне «І», тому його використовують для об'єднання умов у Pandas.

### 3. У чому різниця між формулою та `.apply()`?

Обчислюваний стовпець, створений формулою, виконує одну й ту саму математичну операцію над усіма значеннями стовпця. 
Такий спосіб є векторизованим і зазвичай ефективніший.

`.apply()` дозволяє застосувати власну функцію до кожного значення. 
Його зручно використовувати для складнішої логіки, наприклад для визначення пори року за номером місяця.

```python
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

cloudiness = [78, 74, 68, 58, 50, 44, 40, 42, 52, 66, 76, 82]

wide = pd.DataFrame({
    "місто": ["Київ"],
    "Січ": [78],
    "Лют": [74],
    "Бер": [68],
    "Кві": [58],
    "Тра": [50],
    "Чер": [44],
    "Лип": [40],
    "Сер": [42],
    "Вер": [52],
    "Жов": [66],
    "Лис": [76],
    "Гру": [82]
})

wide
```

```python
wide.shape
```

```python
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

tidy
```

```python
tidy.shape
```

місто
Номінальна шкала, тому що назви міст лише позначають категорії і не мають природного порядку.

місяць
У межах цього завдання порядкова шкала, тому що місяці можна розташувати в певному порядку протягом року.

хмарність
Шкала відношень, тому що це кількісна величина у відсотках, для якої існує природний нуль і можна порівнювати відношення значень.

```python
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

back
```

# Письмові відповіді

### 1. Чому стовпець «місто» — номінальна шкала, а «хмарність» — шкала відношень?

Стовпець «місто» має номінальну шкалу, тому що назви міст є категоріями, між якими немає числового порядку або значення «більше/менше». Стовпець «хмарність» має шкалу відношень, тому що це кількісна величина у відсотках, яка має природний нуль (0%) і для якої можна порівнювати відношення значень.

### 2. Чому wide-таблиця з місяцями в стовпцях не є tidy?

У wide-таблиці кожен місяць є окремим стовпцем, тому одна змінна «місяць» фактично розподілена між багатьма назвами стовпців. Після застосування `melt()` назви місяців перетворюються на значення одного стовпця «місяць», а значення хмарності — на стовпець «хмарність». У результаті кожен місяць представлений окремим рядком.

### 3. Чим CSV відрізняється від JSON?

CSV зберігає інформацію у вигляді таблиці: рядки складаються зі значень, розділених спеціальним символом, наприклад комою. JSON зберігає дані у структурі об'єктів і масивів та може краще представляти складні або вкладені дані. Для цієї таблиці обидва формати дозволяють зберегти однакову інформацію, але записують її по-різному.
